### Setup pymssql package

```python
import pandas as pd
import os
import requests
import json
```

```python
BASKETBALL_DB_API="https://k8plex-veo.vo.elte.hu/notebook/report/wfct0p-basketballdbapi/"
    
def test_health_endpoint() -> None:
    response = requests.get(f"{BASKETBALL_DB_API}/health", timeout=10)
    assert response.status_code == 200

    payload = response.json()
    return payload
 
def show_api_docs() -> None:    
    print(f"Visit the {BASKETBALL_DB_API}/docs for proper documentation")
    

def show_api_paths() -> str:    
    response = requests.get(f"{BASKETBALL_DB_API}/paths", timeout=10)

    payload = response.json()
    return payload

def execute_query(query="SELECT 1 AS one") -> pd.DataFrame:
    response = requests.get(
        f"{BASKETBALL_DB_API}/query",
        params={"query": query},
        timeout=20,
    )
    try:
        assert response.status_code == 200
    except:
        print("ERROR!")
        print(response.json()['detail'])
        return

    payload = response.json()
    df = pd.DataFrame(columns=payload['columns'], data=payload['data'])
    
    return df
    
```
